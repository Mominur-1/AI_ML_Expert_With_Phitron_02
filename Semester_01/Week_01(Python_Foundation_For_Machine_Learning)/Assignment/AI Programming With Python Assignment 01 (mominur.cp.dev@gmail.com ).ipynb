**Question : 1** **(20 Marks)**   
Given a list of integers, extract all numbers that are divisible by 3 but not divisible by 5.

**Input Format:**   
Space-separated integers

**Output Format:**   
A list of numbers satisfying the condition

**Example:**   
Input:   
3 5 9 10 15 18 20 21

Output:   
[3, 9, 18, 21]





```python
input_str = input()
numbers_str = input_str.split()

# Convert strings to integers using a list comprehension
numbers = [int(s) for s in numbers_str]

result = list(filter(lambda x: x % 3 == 0 and x % 5 != 0, numbers))

print(result)
```

**Question : 2** **(20 Marks)**   
Write a function compute(a, b) that:
Returns (a² + b²) / (a - b)
Use exception handling to manage:
Division by zero

**Input Format:**   
Two values

**Output Format:**   
Computed result OR error message

**Example:**   
Input:   
5 5

Output:   
Error: Division not possible

```python
def compute(a, b):
    try:
        result = (a**2 + b**2) / (a - b)
    # except Exception as e:
    #   print(e.with_traceback)
    except ZeroDivisionError:
        return "Error: Division not possible"
    else:
        return result

numbers = list(map(int, input().split()))

print(compute(numbers[0], numbers[1]))
```

**Question : 3** **(20 Marks)**   
Write a Python program that reads a text file and finds the most frequently occurring word.

**Requirements:**

*   Open and read data from a file (Example: data.txt)
*   Ignore case sensitivity ("Python" and "python" as the same)
*   Ignore punctuation marks
*   If multiple words have the same highest frequency, print any one of them

**Input Format:**   
A text file containing multiple lines of text

**Output Format:**   
The most frequent word

**Example File Content (data.txt):**   
Python is easy to learn.
Python is powerful and Python is popular.

**Output:**   
python


```python
# --- Sample Input and Test Setup --- #
sample_filename = "data.txt"
# Creating the file with the content from the example requirement
sample_content = "Python is easy to learn.\nPython is powerful and Python is popular."

import string

with open(f"./sample_data/{sample_filename}", "w") as file:
    file.write(sample_content)

# --- Solution logic --- #

with open(f"./sample_data/{sample_filename}", "r") as file:
    text = file.read()


# Remove punctuation marks

# translator = str.maketrans("xyz", "123", "abc") Mapping x=1, y=2, z=3, a=None, b=npne, c=None
# text.translate(translator) replace all mapping element in text file
translator = str.maketrans("", "", string.punctuation)
text = text.translate(translator)

text = text.lower()
words = text.split()

most_frequent = ""
mx_count = 0
word_count = {}
for word in words:
    word_count[word] = word_count.get(word, 0) + 1
    if word_count[word] > mx_count: mx_count, most_frequent = word_count[word], word


# most_frequent = max(word_count, key=word_count.get)

print(most_frequent)
```

**Question : 4** **(20 Marks)**   
Write a program that converts a sentence into a list representing the length of each word.Ignore extra spaces

**Input Format:**   

A single line sentence

**OutputFormat:**   
A list of integers representing word lengths

**Example:**   
Input:   
Machine learning is fun

```python
words_length = list(map(lambda x: len(x), input().split()))
print(words_length)
```

**Question : 5** **(20 Marks)**   
A number is called a Mirror Number if reversing its digits results in the same number.

Write a program to check whether a number is a mirror number.

**Input Format:**   
An integer

**Output Format:**   
"Mirror Number" or "Not Mirror Number"

**Example:**   
Input:   
121

Output:   
Mirror Number


```python
orginal_num = input()
reversed_num = orginal_num[::-1]

status = "Mirror" if orginal_num == reversed_num else "Not Mirror"
print(f"{status} Number")
```
