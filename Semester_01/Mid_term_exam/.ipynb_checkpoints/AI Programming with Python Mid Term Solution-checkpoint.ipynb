## Starter Code (Run First)

```python
# You must run this code and use this dataset for all questions

import numpy as np
import pandas as pd

np.random.seed(24)

num_employees = 50 # you can increase this if needed

# Employee data
emp_ids = np.arange(101, 101 + num_employees)
emp_ages = np.random.randint(20, 55, num_employees)
emp_income = np.random.randint(35000, 95000, num_employees)

# Fixed set of roles (repetition allowed)
roles_list = [
"HR Manager",
"Software Developer",
"Accountant",
"Marketing Executive",
"Sales Executive",
"Backend Developer",
"HR Executive",
"Financial Analyst",
"Digital Marketer",
"Operations Manager"
]

emp_roles = np.random.choice(roles_list, num_employees)

# Create DataFrame
data = pd.DataFrame(
    {
    "emp_id": emp_ids,
    "emp_age": emp_ages,
    "emp_income": emp_income,
    "role": emp_roles
    }
)

# Save dataset
data.to_csv("staff_data.csv", index=False)
```

## 1) NumPy — 5 Questions

### Q1. Use NumPy arrays , the emp_ages and emp_income. Print both arrays and display:

```python
# ● data type
# ● number of dimensions
# ● shape
# ● total number of elements

import numpy as np


data = np.genfromtxt('staff_data.csv', delimiter=',', names=True, skip_header=False , dtype=None, encoding='utf-8')

print('\n')

print("emp_ages all properties :")
print(f" data type : {data['emp_age'].dtype}")
print(f" number of dimensions : {data['emp_age'].ndim}")
print(f" sahpe : {data['emp_age'].shape}")
print(f" total number of elements : {data['emp_age'].size}")
print("\n")

print("emp_income all properties :")
print(f" data type : {data['emp_income'].dtype}")
print(f" number of dimensions : {data['emp_income'].ndim}")
print(f" sahpe : {data['emp_income'].shape}")
print(f" total number of elements : {data['emp_income'].size}")
```

### Q2. Using NumPy functions:


```python
# ● Find the maximum and minimum salary
# ● Calculate the average salary
# ● Calculate the average age
# Print all results.

print(f"Maximum salary : {data['emp_income'].max()}")
print(f"Minimum salary : {data['emp_income'].min()}")
print(f"Average salary : {data['emp_income'].mean()}")
print(f"Average age : {data['emp_age'].mean()}")
```

### Q3. From the emp_ages array(Filter)

```python
# Extract employees who:
# ● are aged between 25 and 40,
# AND
# Emp_salary : have salary greater than 50,000
# Then :
# ● Print the filtered values
# ● Count how many employees satisfy this condition

filtered = (data['emp_age'] >= 25) & (data['emp_age'] <= 40) & (data['emp_income'] > 50000)

print(f"The filtered values :\n {data[filtered]}\n")
print(f"Satisfyed employe :\n{len(data[filtered])}")
```

### Q4. Create a new NumPy array where each employee’s age is adjusted based on:

```python
# ● Add 4 years if age is less than the average age
# ● Add 2 years if age is greater than or equal to the average age
# Do NOT modify the original array.
# Print:
# ● original array
# ● updated array
# ● average age used for condition

Ave_age = data['emp_age'].mean()

updated_data = data.copy()
updated_data['emp_age'] = np.where(data['emp_age'] < Ave_age, data['emp_age']+4, data['emp_age']+2)

print(f"original array :\n {data}\n")
print(f"updated array :\n {updated_data}\n")
print(f"average age used for condition : {Ave_age}")
```

### Q5. Using NumPy:

```python
# ● Assume all salaries of Tech roles increase by 8%
# ● Calculate the new total salary expense
# ● Find the overall increase in expense after the raise
# Print both old and new totals along with the difference.

tech_role = np.char.find(data['role'], 'Developer') != -1

old_total_expense = data['emp_income'].sum()

p = 8/100

new_salaries = np.where(tech_role, (data['emp_income']+(data['emp_income']*p)) , data['emp_income'])

new_total_expense = new_salaries.sum()
expense_difference = new_total_expense - old_total_expense

print(f"Old Total Salary Expense : {old_total_expense:,.2f}")
print(f"New Total Salary Expense : {new_total_expense:,.2f}")
print(f"Overall Increase Amount  : {expense_difference:,.2f}")
tech_role
```

### Q7. Filter the dataset to include only:

```python
# ● Select employees whose age is between 25 and 40
# ● Sort them by emp_id
# ● Display the last 3 rows of the result

import pandas as pd

filtered_df = df[['emp_id', 'emp_age', 'role']]

conditioned_df = filtered_df[(filtered_df['emp_age'] >= 25) & (filtered_df['emp_age'] <= 40)]


sorted_df = conditioned_df.sort_values('emp_id')

print(sorted_df.tail(3))
```

### Q9. Sort the DataFrame by:

```python
# ● emp_age (ascending)
# ● and then by emp_salary (descending)
# Display the top 5 employees after sorting, including:
# ● emp_id
# ● role
# ● emp_salary
sorted_df = df.sort_values(['emp_age', 'emp_income'], ascending=[1,0])

print(f"Top 5 Employees:\n{sorted_df[['emp_id', 'role', 'emp_income']].head(5)}")
```

### Q10. Using .loc:

```python
# ● Increase salary by 12% for "Software Developer" and "Backend Developer"
# ● Increase salary by 5% for all other roles
# Then:
# ● Print the dataframe that includes role , previous_salary , new_salary , increase_of_salary

df['previous_salary'] = df['emp_income']
tech_role = (df["role"] == "Software Developer") | (df["role"] == "Backend Developer")
# tech_role = df["role"].isin(["Software Developer", "Backend Developer"])

# P Increase x% meaning = P + (P*(x/100))
# => P(1+(x/100))

df.loc[tech_role, 'new_salary'] = df.loc[tech_role, 'previous_salary']*(1+(12/100))
df.loc[~tech_role, 'new_salary'] = df.loc[~tech_role, 'previous_salary']*(1+(5/100))

df.loc[:, 'increase_of_salary'] = df.loc[:, 'new_salary'] - df.loc[:, 'previous_salary']

print(df)
```
