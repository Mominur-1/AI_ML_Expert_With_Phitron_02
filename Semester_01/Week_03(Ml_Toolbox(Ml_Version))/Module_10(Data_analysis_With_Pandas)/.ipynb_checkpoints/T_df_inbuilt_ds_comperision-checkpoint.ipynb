```python
import pandas as pd
import sys

print("=" * 80)
print("PANDAS DATAFRAME INPUT FORMATS COMPARISON")
print("=" * 80)

# ============================================================================
# 1. LIST OF LISTS
# ============================================================================
print("\n\n1️⃣  LIST OF LISTS")
print("-" * 80)

my_list = [['Alice', 25], ['Bob', 30], ['Charlie', 28]]
list_df = pd.DataFrame(my_list)

print("Input Structure:")
print(f"  my_list = {my_list}")
print(f"  type: {type(my_list)}")
print(f"  Memory: {sys.getsizeof(my_list)} bytes")

print("\nDataFrame Output:")
print(list_df)

print("\nColumn Names Issue:")
print(f"  Default columns: {list_df.columns.tolist()}")
print("  ❌ Column names are auto-generated (0, 1, 2...)")

print("\nFixed Version (with column names):")
list_df = pd.DataFrame(my_list, columns=['Name', 'Age'])
print(list_df)


# ============================================================================
# 2. TUPLE OF TUPLES
# ============================================================================
print("\n\n2️⃣  TUPLE OF TUPLES")
print("-" * 80)

my_tuple = (('Alice', 25), ('Bob', 30), ('Charlie', 28))
tuple_df = pd.DataFrame(my_tuple)

print("Input Structure:")
print(f"  my_tuple = {my_tuple}")
print(f"  type: {type(my_tuple)}")
print(f"  Memory: {sys.getsizeof(my_tuple)} bytes")

print("\nDataFrame Output:")
print(tuple_df)

print("\nColumn Names Issue:")
print(f"  Default columns: {tuple_df.columns.tolist()}")
print("  ❌ Same issue as list of lists")

print("\nFixed Version (with column names):")
tuple_df = pd.DataFrame(my_tuple, columns=['Name', 'Age'])
print(tuple_df)


# ============================================================================
# 3. DICTIONARY OF LISTS
# ============================================================================
print("\n\n3️⃣  DICTIONARY OF LISTS")
print("-" * 80)

my_dict = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 28]
}
dict_df = pd.DataFrame(my_dict)

print("Input Structure:")
print(f"  my_dict = {my_dict}")
print(f"  type: {type(my_dict)}")
print(f"  Memory: {sys.getsizeof(my_dict)} bytes")

print("\nDataFrame Output:")
print(dict_df)

print("\nColumn Names:")
print(f"  ✅ Columns automatically from dict keys: {dict_df.columns.tolist()}")


# ============================================================================
# 4. LIST OF DICTIONARIES
# ============================================================================
print("\n\n4️⃣  LIST OF DICTIONARIES")
print("-" * 80)

my_data = [
    {'Name': 'Alice', 'Age': 25, 'City': 'New York'},
    {'Name': 'Bob', 'Age': 30, 'City': 'Paris'},
    {'Name': 'Charlie', 'Age': 28}
]
data_df = pd.DataFrame(my_data)

print("Input Structure:")
print(f"  my_data = {my_data}")
print(f"  type: {type(my_data)}")
print(f"  Memory: {sys.getsizeof(my_data)} bytes")

print("\nDataFrame Output:")
print(data_df)

print("\nColumn Names:")
print(f"  ✅ Columns from dict keys: {data_df.columns.tolist()}")

print("\nMissing Data Handling:")
print("  ✅ Missing 'City' for Charlie handled as NaN")


# ============================================================================
# COMPARISON TABLE
# ============================================================================
print("\n\n" + "=" * 80)
print("DETAILED COMPARISON TABLE")
print("=" * 80)

comparison_data = {
    'Feature': [
        'Auto Column Names',
        'Handles Missing Data',
        'Readability',
        'Memory Efficient',
        'Mutable',
        'Conversion Effort',
        'Self-Documenting',
        'Best for Production'
    ],
    'List of Lists': [
        '❌ No',
        '❌ No',
        '❌ Poor',
        '✅ Yes',
        '✅ Yes',
        '❌ Medium',
        '❌ No',
        '❌ No'
    ],
    'Tuple of Tuples': [
        '❌ No',
        '❌ No',
        '❌ Poor',
        '✅ Yes',
        '❌ No',
        '❌ Medium',
        '❌ No',
        '❌ No'
    ],
    'Dict of Lists': [
        '✅ Yes',
        '❌ No',
        '✅ Good',
        '❌ More',
        '✅ Yes',
        '✅ Easy',
        '✅ Yes',
        '✅ YES ⭐'
    ],
    'List of Dicts': [
        '✅ Yes',
        '✅ Yes',
        '✅ Excellent',
        '❌ More',
        '✅ Yes',
        '✅ Easy',
        '✅ Yes',
        '✅ YES ⭐'
    ]
}

comparison_df = pd.DataFrame(comparison_data)
# Create a formatter dict to left-align string data in each column
formatters = {col: lambda x: f"{x:<25}" for col in comparison_df.columns}

# Print with left justification for headers and our custom formatters
print(comparison_df.to_string(index=False, justify='left', formatters= formatters))


# ============================================================================
# DETAILED PROS & CONS
# ============================================================================
print("\n\n" + "=" * 80)
print("DETAILED PROS & CONS")
print("=" * 80)

print("\n1️⃣  LIST OF LISTS")
print("-" * 80)
print("PROS:")
print("  ✅ Memory efficient")
print("  ✅ Simple structure")
print("  ✅ Fast to create")
print("\nCONS:")
print("  ❌ Must specify column names manually")
print("  ❌ Cannot handle missing data")
print("  ❌ Poor readability (must remember index)")
print("  ❌ Not self-documenting")

print("\n2️⃣  TUPLE OF TUPLES")
print("-" * 80)
print("PROS:")
print("  ✅ Immutable (thread-safe)")
print("  ✅ Memory efficient")
print("  ✅ Can be used as dict key")
print("\nCONS:")
print("  ❌ Must specify column names manually")
print("  ❌ Cannot handle missing data")
print("  ❌ Poor readability")
print("  ❌ Rarely used in practice")

print("\n3️⃣  DICTIONARY OF LISTS")
print("-" * 80)
print("PROS:")
print("  ✅ Auto column names from keys")
print("  ✅ Self-documenting")
print("  ✅ Readable")
print("  ✅ No conversion needed for DataFrame")
print("  ✅ Standard format for APIs")
print("\nCONS:")
print("  ❌ All columns must have same length")
print("  ❌ Cannot handle missing data per record")
print("  ❌ More memory than list of lists")
print("  ⚠️  Less flexible than list of dicts")

print("\n4️⃣  LIST OF DICTIONARIES")
print("-" * 80)
print("PROS:")
print("  ✅ Auto column names from dict keys")
print("  ✅ Handles missing data (NaN)")
print("  ✅ Most readable & self-documenting")
print("  ✅ Easy to modify individual records")
print("  ✅ Flexible structure")
print("  ✅ Standard for JSON/APIs")
print("  ✅ Best practice in production")
print("\nCONS:")
print("  ❌ More memory (stores keys)")
print("  ❌ Slightly slower (hash lookups)")
print("  ⚠️  Slightly more verbose")


# ============================================================================
# TIME & SPACE COMPLEXITY
# ============================================================================
print("\n\n" + "=" * 80)
print("TIME & SPACE COMPLEXITY")
print("=" * 80)

complexity_data = {
    'Operation': [
        'Creation',
        'Access Element',
        'Add Row',
        'Memory Usage',
        'DataFrame Conversion'
    ],
    'List of Lists': [
        'O(n)',
        'O(1)',
        'O(1)',
        'Low',
        'O(n) - needs columns'
    ],
    'Tuple of Tuples': [
        'O(n)',
        'O(1)',
        'O(n) immutable',
        'Low',
        'O(n) - needs columns'
    ],
    'Dict of Lists': [
        'O(n)',
        'O(1)',
        'O(n) grow all lists',
        'Medium',
        'O(n) direct'
    ],
    'List of Dicts': [
        'O(n)',
        'O(1)',
        'O(1)',
        'High',
        'O(n) direct'
    ]
}

complexity_df = pd.DataFrame(complexity_data)

# Create a formatter dict to left-align string data in each column
formatters = {col: lambda x: f"{x:<25}" for col in comparison_df.columns}

# Print with left justification for headers and our custom formatters
print(complexity_df.to_string(index=False, justify='left', formatters= formatters))

# ============================================================================
# PRACTICAL EXAMPLES
# ============================================================================
print("\n\n" + "=" * 80)
print("PRACTICAL EXAMPLES & USE CASES")
print("=" * 80)

print("\n📊 CASE 1: Data from API/JSON")
print("-" * 80)
print("Source: REST API returns JSON")
print("\nNATURAL FORMAT: List of Dictionaries")
json_data = [
    {'id': 1, 'name': 'Alice', 'email': 'alice@example.com'},
    {'id': 2, 'name': 'Bob', 'email': 'bob@example.com'},
    {'id': 3, 'name': 'Charlie', 'email': 'charlie@example.com'}
]
json_df = pd.DataFrame(json_data)
print(json_df)

print("\n📊 CASE 2: Data from Database")
print("-" * 80)
print("Source: SQL Query returns rows")
print("\nNATURAL FORMAT: List of Dictionaries (via ORM)")
db_data = [
    {'product': 'Laptop', 'price': 1000, 'stock': 5},
    {'product': 'Mouse', 'price': 25, 'stock': 150},
    {'product': 'Keyboard', 'price': 75, 'stock': 80}
]
db_df = pd.DataFrame(db_data)
print(db_df)

print("\n📊 CASE 3: Data for ML Training")
print("-" * 80)
print("Source: Numpy arrays or calculations")
print("\nNATURAL FORMAT: Dictionary of Lists")
ml_data = {
    'feature1': [1.2, 2.3, 3.1, 4.5],
    'feature2': [5.1, 6.2, 7.3, 8.4],
    'target': [0, 1, 0, 1]
}
ml_df = pd.DataFrame(ml_data)
print(ml_df)

print("\n📊 CASE 4: Numerical Matrix Data")
print("-" * 80)
print("Source: Matrix computation or coordinates")
print("\nNATURAL FORMAT: List of Lists OR Dict of Lists")
matrix_data = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]
matrix_df = pd.DataFrame(matrix_data, columns=['X', 'Y', 'Z'])
print(matrix_df)


# ============================================================================
# CONVERSION BETWEEN FORMATS
# ============================================================================
print("\n\n" + "=" * 80)
print("CONVERSION BETWEEN FORMATS")
print("=" * 80)

print("\n1️⃣  List of Lists → Dictionary of Lists")
print("-" * 80)
list_data = [['Alice', 25], ['Bob', 30], ['Charlie', 28]]
dict_from_list = {
    'Name': [row[0] for row in list_data],
    'Age': [row[1] for row in list_data]
}
print(f"Input:  {list_data}")
print(f"Output: {dict_from_list}")

print("\n2️⃣  Dictionary of Lists → List of Dictionaries")
print("-" * 80)
dict_data = {'Name': ['Alice', 'Bob'], 'Age': [25, 30]}
list_of_dict = [dict(zip(dict_data.keys(), values)) 
                for values in zip(*dict_data.values())]
print(f"Input:  {dict_data}")
print(f"Output: {list_of_dict}")

print("\n3️⃣  Tuple of Tuples → Dictionary of Lists")
print("-" * 80)
tuple_data = (('Alice', 25), ('Bob', 30))
dict_from_tuple = {
    'Name': [row[0] for row in tuple_data],
    'Age': [row[1] for row in tuple_data]
}
print(f"Input:  {tuple_data}")
print(f"Output: {dict_from_tuple}")


# ============================================================================
# RECOMMENDATION & BEST PRACTICES
# ============================================================================
print("\n\n" + "=" * 80)
print("🏆 RECOMMENDATION & BEST PRACTICES")
print("=" * 80)

print("\n✅ FOR PRODUCTION CODE USE:")
print("-" * 80)
print("\n#1 CHOICE: LIST OF DICTIONARIES")
print("   Why:")
print("   • Handles missing data (NaN)")
print("   • Self-documenting (keys describe data)")
print("   • Flexible (different keys per record)")
print("   • Direct conversion to DataFrame")
print("   • Standard for APIs/JSON")
print("   • Recommended by Pandas documentation")
print("\n   When to use:")
print("   • Data from APIs/JSON")
print("   • Database queries (ORM objects)")
print("   • Real-world structured data")
print("   • Production applications")

print("\n#2 CHOICE: DICTIONARY OF LISTS")
print("   Why:")
print("   • Good balance of readability & efficiency")
print("   • Clean column names")
print("   • Direct DataFrame conversion")
print("   • Slightly more memory efficient")
print("\n   When to use:")
print("   • Machine learning feature sets")
print("   • All records have identical structure")
print("   • Data manipulation heavy code")
print("   • Scientific computing")

print("\n#3 CHOICE: LIST OF LISTS")
print("   Why:")
print("   • Memory efficient")
print("   • Fast creation")
print("   • Simple structure")
print("\n   When to use:")
print("   • Pure numerical data (matrices)")
print("   • All records identical")
print("   • Memory is critical")
print("   • Quick scripts/exploration")
print("   • Rarely recommended")

print("\n❌ AVOID: TUPLE OF TUPLES")
print("   • No advantage over list of lists")
print("   • Less flexible")
print("   • Rarely used in practice")


# ============================================================================
# SUMMARY TABLE
# ============================================================================
print("\n\n" + "=" * 80)
print("📋 QUICK SUMMARY")
print("=" * 80)

summary = """
Format                 | Use for           | Pros              | Cons
─────────────────────────────────────────────────────────────────────────
List of Lists          | Simple data       | Memory efficient  | No column names
Tuple of Tuples        | Immutable data    | Thread-safe       | Poor for DataFrames
Dict of Lists          | ML/Numerical      | Clean structure   | Fixed columns
List of Dicts ⭐       | Production code   | Flexible & Clean  | More memory
"""
print(summary)

print("\n🎯 FINAL ANSWER: Use LIST OF DICTIONARIES for real-world code!")
print("   It's the best balance of readability, flexibility, and usability.")
print("=" * 80)
```
