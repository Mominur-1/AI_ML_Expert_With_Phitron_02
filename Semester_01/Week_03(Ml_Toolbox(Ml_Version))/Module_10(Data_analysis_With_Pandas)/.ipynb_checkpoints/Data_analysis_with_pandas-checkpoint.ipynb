## Transforming to df from inbuilt datas

```python
my_list = [['Alice', 25], ['Bob', 30], ['Charlie', 28]]

list_df = pd.DataFrame(my_list)

print(my_list)
print(type(my_list))

print(list_df)
print(type(list_df))
```

```python
list_df = pd.DataFrame(my_list, columns =['Name', 'Age'], index=[1,2,3])

print(type(list_df))
print(list_df)
```

```python
list_df = pd.DataFrame(my_list, columns =['Name', 'Age'], index=range(1, len(my_list)+1, 1))

print(type(list_df))
print(list_df)
```

```python
my_tuple = (('Alice', 25), ('Bob', 30), ('Chaelie', 28))

tuple_df = pd.DataFrame(my_tuple)

print(my_tuple)

tuple_df
```

```python
my_dict = {
    'Name' : ['Alice', 'Bob', 'Charlie'],
    'Age' : [25, 30, 28]
}

dict_df = pd.DataFrame(my_dict)

print(my_dict)
dict_df
```

```python
my_data = [
    {'Name':'Alice', 'Age':25, 'City': 'New York'},
    {'Name':'Bob', 'Age':30, 'City': 'Paris'},
    {'Name':'Charlie', 'Age':28}
]

data_df = pd.DataFrame(my_data)

data_df
```

## Changing  index and columns using iloc

```python

df_idx = df.set_index(['StudentID', 'FullName'])

df_idx
```

```python
# for loc ending is inclusive, for iloc ending is exclusive and for both starting is inclusive
# in loc we need actual row and column name with case sensitive
# in iloc we only use numeric number, row or columns(string, number or etc)

print(df_idx.iloc[:,1:3])
# iloc = integer location

df_idx.loc['PH1020']
# loc = location
```

```python
df.rename(columns={'FullName':'Full Name', 'Algorithm Marks':'Algo Marks'})
# its return copy. if use inplace=True then it edit orginal
```

```python
# row deletion

df.drop(0, inplace=True)
df
```
