Problem 1: Create and Write to a File

Write a function create_file(filename, content) that creates a new text file with the given filename and writes the content into it. Call it using positional arguments.

📥: Sample Input:  
"notes.txt", "Hello, this is my first file."  
📤: Sample Output (notes.txt will contain):  
Hello, this is my first file.



```python
def create_file(filename, content):
    """
    Creates a new text file with the given filename and writes the content into it.
    """
    with open(f"./sample_data/{filename}", "w") as file:
        file.write(content)

# Sample Input
filename = "notes.txt"
content = "Hello, this is my first file."

# Call the function using positional arguments
create_file(filename, content)

print(f"File '{filename}' created successfully.")

# Optional: Verify the content of the file
with open(f"./sample_data/{filename}", "r") as file:
    print(f"Content of '{filename}': {file.read()}")
```

Problem 2: Read Lines with Line Numbers

Write a function read_lines(filename) that reads a text file and prints each line with its line number.  

📥 Sample Input (file: data.txt):  
Apple  
Banana  
Cherry  
📤 Sample Output:  
1: Apple  
2: Banana  
3: Cherry  


```python
def read_lines(filename):
    """
    Reads a text file and prints each line with its line number using a manual counter.
    """
    try:
        with open(f"./sample_data/{filename}", "r") as file:
            line_number = 1
            for line in file:
                print(f"{line_number}: {line.strip()}")
                line_number += 1
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")
    except Exception as e:
        print(f"An error occurred: {e}")

# --- Sample Input and Test --- #

# Create a sample data.txt file for testing
sample_filename = "data.txt"
sample_content = "Apple\nBanana\nCherry"

with open(f"./sample_data/{sample_filename}", "w") as file:
    file.write(sample_content)

print(f"Created '{sample_filename}' with sample content.\n")

# Call the function with the sample file
read_lines(sample_filename)
```

Problem 3: Append Log Entries to a File

Given a filename log.txt, use file append mode to add a new log entry each time the function add_log(message) is called. Each message should appear on a new line.  

📥 Sample Input:  
add_log("User logged in")  
add_log("File uploaded")  
📤 Sample Output (log.txt will contain):  
User logged in  
File uploaded  

```python
def add_log(message):
    filename = "log.txt"
    try:
        with open(f"./sample_data/{filename}", "a") as file:
            file.write(f"{message}\n") # Using an f-string here
        print(f"Log entry added to '{filename}' using f-string: {message}")
    except Exception as e:
        print(f"An error occurred while writing to log file: {e}")

# --- Sample Input and Test for f-string version --- #

# Clear the log file for consistent testing
with open("./sample_data/log.txt", "w") as file:
    file.write("")

add_log("user logged in")
add_log("file uploaded")

print("\n--- Content of log.txt ---")
try:
    with open("./sample_data/log.txt", "r") as file:
        print(file.read())
except FileNotFoundError:
    print("log.txt not found.")
except Exception as e:
    print(f"An error occurred while reading log file: {e}")
```

Problem 4: Read from a Specific File Pointer Position

Write a function read_from_position(filename, position) that uses the file pointer to start reading a file from a given byte position and prints the remaining content.  

📥 Sample Input:  
filename = "hello.txt"   # contains: "Hello, World!"  
position = 7  
📤 Sample Output:  
World!





```python
def read_from_position(filename, position):
    """
    Reads a text file from a given byte position and prints the remaining content.
    """
    try:
        with open(f"./sample_data/{filename}", "r") as file:
            file.seek(position)
            print(file.read())
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found.")
    except Exception as e:
        print(f"An error occurred: {e}")

# --- Sample Input and Test --- #

# Create a sample 'hello.txt' file for testing
sample_filename = "hello.txt"
sample_content = "Hello, World!"

with open(f"./sample_data/{sample_filename}", "w") as file:
    file.write(sample_content)

print(f"Created '{sample_filename}' with content: '{sample_content}'\n")

# Call the function with the sample file and position
position = 7
print(f"Reading '{sample_filename}' from position {position}:")
read_from_position(sample_filename, position)
```

Problem 5: ⭐ Conceptual Session — Safe File Read with Full Exception Handling

Print file content if successful (else block)
Handle the case where file does not exist (except block)
Always print "Operation complete." at the end (finally block)  



📥 Sample Input 1:

```python
from types import TracebackType
def safe_read(filename):
    """
    Tries to open and read a file. Uses
    try, except, else, and finally blocks for robust file handling.
    """
    try:
        with open(f"./sample_data/{filename}", "r") as file:
            content = file.read()
    # except Exception as e:
    #   print(e.with_traceback)
    except FileNotFoundError:
        print("Error: File not found.")
    else:
        print(content)
    finally:
        print("Operation complete.")

# --- Sample Input 1 and Test ---
print("\n--- Testing missing.txt ---")
safe_read("missing.txt")

# --- Sample Input 2 and Test ---
# Ensure hello.txt exists from a previous problem or create it
# This part is just for testing consistency
# sample_filename = "hello.txt"
# sample_content = "Hello, World!"
# with open(f"./sample_data/{sample_filename}", "w") as file:
#    file.write(sample_content)

print("\n--- Testing hello.txt ---")
safe_read("hello.txt")
```
