## Practice Problem (Array dimension using numpy)

### 1 . Create a 1D, 2D, and 3D NumPy array and print their dimensions and shapes.

```python
import numpy as np

# 1d array

array_1d = np.array([1, 2, 3, 4])

# 2d arraay
array_2d = np.array(
    [
        [1, 2, 3, 4], # 1st 1d
        [1, 2, 3, 4] # 2nd 1d
    ]
)

array_2d1 = np.array([array_1d, array_1d])

# 3d array
array_3d = np.array(
    [
        [[1, 2, 3, 4], [1, 2, 3, 4]], #1st 2d
        [[1, 2, 3, 4], [1, 2, 3, 4]] #2nd 2d
    ]
)

array_3d1 = np.array([array_2d, array_2d])

print(" 1 D array")
print(array_1d.shape)
print(array_1d.ndim)
print(array_1d, "\n")

print(" 2 D array")
print(array_2d.shape)
print(array_2d.ndim)
print(array_2d1.shape)
print(array_2d1.ndim)
print(array_2d, "\n\n", array_2d1, "\n\n")

print(" 3 D array")
print(array_3d.shape)
print(array_3d.ndim)
print(array_3d1.shape)
print(array_3d1.ndim)
print(array_3d, "\n\n", array_3d1)
```
