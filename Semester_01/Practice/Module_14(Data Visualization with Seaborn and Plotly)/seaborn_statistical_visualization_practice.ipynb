```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
sale = pd.read_csv('sales.csv')
sale
```

```python
student = pd.read_csv('students.csv') 
student
```

### Using sales, draw a lineplot of date vs sales

```python
sns.lineplot(data=sale, x='date', y='sales')
plt.title(' lineplot of date vs sales')
plt.show()
```

### From students, plot hours_studied vs marks and color by gender

```python
# line plot
sns.scatterplot(data=student, x='hours_studied', y='marks', hue='gender')
plt.title('plot hours_studied vs marks and color by gender')
plt.show() 
```

### Create a grid of scatterplots (hours_studied vs marks) faceted by subject. 

```python
# scatterplots using seaborn
sns.relplot(
    kind='scatter',
    data=student, 
    x="hours_studied", 
    y="marks", 
    hue='gender',
    col ='subject'
)
```

### Plot a histogram of marks with a KDE overlay

```python
# Create histogram with KDE overlay
sns.histplot(
    data=student, 
    x="marks", 
    kde=True,
    bins=10
)

plt.title("Distribution of Marks with KDE")
plt.show()
```

```python
# 1. Plot the histogram with its own color
sns.histplot(
    data=student, 
    x="marks",
    stat='density', 
    color="skyblue",
    bins=10
)

# 2. Plot the KDE separately on top with a different color
sns.kdeplot(
    data=student, 
    x="marks", 
    color="red",
    linewidth=2
)

plt.title("Histogram with Custom KDE Color")
plt.show()
```

### Show the KDE curve for hours_studied.


```python
sns.kdeplot(
    data=student,
    x='hours_studied'
)

plt.title('The KDE curve for hours_studied')
plt.show()
```

### Display the count of records per subject.

```python
# 1. Get the count of records per subject using pandas
subject_counts = student["subject"].value_counts()
print(subject_counts)

# 2. Plot the counts using Seaborn
sns.countplot(
    data=student, 
    x="subject",
    hue='gender'
)

plt.title("Number of Students per Subject")
plt.show()
```

### Draw a regression line for hours_studied and marks

```python
sns.regplot(
    data=student,
    x='hours_studied',
    y='marks',
    color="green",
    scatter_kws={"alpha": 0.7},
    line_kws={"color": "red"}
)
```

### Create a pairplot for students' dataset.

```python
sns.pairplot(
    student, 
    vars=["hours_studied", "marks", "attendance_percentage", "previous_gpa", "age"], 
    hue="gender", 
    palette="Set1"
)
```
