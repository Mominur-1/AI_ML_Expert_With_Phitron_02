## Add New Column, Check Duplicates and Remove, Handle Missing Values,  Apply Statistical, Histogram, Bar, Scatter, Pie

```python
import pandas as pd
Employee = pd.read_csv('Practice.csv')

Employee
```

### 1. Add New Column and Save File

```python
# Create a new column 'Discounted_Price' with a 10% discount on Employee['Price']
Employee['Discounted_Price'] = Employee['Price'] *(1-0.1)

# Save the modified DataFrame as discounted_products.csv[cite: 2]
Employee.to_csv('discounted_products.csv')
```

### 2. Check Duplicates and Remove Them

```python
# Count duplicate entries in the 'EmployeeID' column
duplicate_count = Employee['EmployeeID'].duplicated().sum()

# Remove duplicates based on 'EmployeeID'
Employee.drop_duplicates(subset=['EmployeeID'], inplace=True)

Employee
```

### 3. Handle Missing Values

```python
# Count missing values
print(f"Count missing values of Age : {Employee['Age'].isna().sum()}")
print(f"Count missing values of Salary : {Employee['Salary'].isna().sum()}")

# Fill missing Age with mean age
Employee.fillna({'Age': Employee['Age'].mean()}, inplace=True)

print(Employee)

Employee.dropna(subset=('Salary'), inplace=True)

Employee
```

### 4. Apply Statistical Functions


```python
# Columns: Marks1, Marks2, Marks3

# Calculate mean, max for each subject
print(f"Calculate mean, max for each for Marks1 : {Employee['Marks1'].mean():.2f}")
print(f"Calculate mean, max for each for Marks2 : {Employee['Marks2'].mean():.2f}")
print(f"Calculate mean, max for each for Marks3 : {Employee['Marks3'].mean():.2f}")

# Compute total marks per student
total_marks = Employee['Marks1'] + Employee['Marks2'] + Employee['Marks3']

print(f"Total marks per student :\n{total_marks}")

total_marks.std()
```

### 5. Histogram, Bar, Scatter, Pie

```python
Employee
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

#### Histogram

```python
# matplot

plt.hist(Employee['Age'])
plt.title('Histogram of Age (Matplotlib)')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

```python
# seaborn

sns.histplot(Employee['Age'])
plt.title('Histogram of Age (Seaborn)')

plt.show()
```

```python
# plotly

fig = px.histogram(Employee, x='Age', title='Histogram of Age (Plotly)')
fig.show()
```

#### Bar Chart

```python
# matplot

plt.bar(Employee['EmployeeName'], Employee['Salary'])
plt.title('Bar Chart of Salary (Matplotlib)')
plt.xlabel('Name')
plt.ylabel('Salary')
plt.xticks(rotation=45)
plt.show()
```

```python
# seaborn

sns.barplot(data=Employee, x='EmployeeName', y='Salary')
plt.title('Bar Chart of Salary (Seaborn)')
plt.xticks(rotation=45)
plt.show()
```

```python
# plotly

fig = px.bar(Employee, x='EmployeeName', y='Salary', title='Bar Chart of Salary (Plotly)')
fig.show()
```

### Scatter Plot

```python
# matplot 

plt.scatter(Employee['Age'], Employee['Salary'])
plt.title('Scatter Plot: Age vs Salary (Matplotlib)')
plt.xlabel('Age')
plt.ylabel('Salary')
plt.show()
```

```python
# seaborn

sns.scatterplot(data=Employee, x='Age', y='Salary')
plt.title('Scatter Plot: Age vs Salary (Seaborn)')
plt.show()
```

```python
# plotly

fig = px.scatter(Employee, x='Age', y='Salary', title='Scatter plot: Age vs Salary (plotly)')
fig.show()
```

### Pie Chart

```python
# matplot

salary_by_name = Employee.groupby('EmployeeName')['Salary'].sum()
plt.pie(salary_by_name, labels=salary_by_name.index, autopct='%1.1f%%')
plt.title('Pie Chart of Employee Names (Matplotlib)')
plt.show()
```

```python
# plotly

salary_by_name = Employee.groupby('EmployeeName')['Salary'].sum().reset_index()

fig = px.pie(salary_by_name, names='EmployeeName', values='Salary', title='Pie Chart of Salary by Employee (Plotly)')

fig.show()
```

### 6. Histogram , Scatter, Correlation

```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

#### Load the dataset from the URL into a pandas DataFrame.

```python
url = "https://drive.google.com/uc?id=1Sy7rA0G9gTmeItVZ-XlugkKF4CPPqvxU"
students = pd.read_csv(url)
students
```

#### Create a scatter plot with Study_Hours on the X-axis and IQ on the Y-axis.

```python
fig = px.scatter(
    students, 
    x='Study_Hour', 
    y='IQ_Score', 
    title='Scatter plot with Study_Hour on the X-axis and IQ on the Y-axis using plotly'
)
fig.show()
```

#### Create a scatter plot with Shoe_Size on the X-Axis and Study_Hours on the Y-axis

```python
plt.scatter(students['Shoe_Size'], students['Study_Hour'])
plt.xlabel('Shoe_Size')
plt.ylabel('Study_Hours')
plt.title(' Scatter plot with Shoe_Size on the X-Axis and Study_Hours on the Y-axis. using matplot')
```

#### Create a histogram of Shoe_Size

```python
fig = px.histogram(students, x='Shoe_Size', title='A histogram of Shoe_Size using plotly')
fig.update_layout(
    xaxis_title='Shoe Size',
    yaxis_title='Frequency'
)
fig.show()
```

```python
# plotly

fig = px.scatter(students, x='Shoe_Size', y='IQ_Score', title='Scatter plot: Shoe_Size vs IQ_Score (plotly)')
fig.show()
```
