### 1. Create a Pandas Series from the list [10, 20, 30, 40, 50]. Print the Series, its index, and data type. Then create a DataFrame with columns "Name", "Age", and "City" containing at least 4 rows of data.

```python
import pandas as pd

# Create a Series from a list
my_list = [10, 20, 30, 40, 50]
my_series = pd.Series(my_list)

print(my_series, '\n')
print(my_series.index)
print(my_series.dtype)


# Method 1: Using dictionary (recommended)
data_dict = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 28, 35],
    'City': ['New York', 'Paris', 'Tokyo', 'London']
}

df = pd.DataFrame(data_dict)

print("\n1. The DataFrame:")
print(df)

print("\n2. DataFrame Info:")
print(f"Type: {type(df)}")
print(f"Shape: {df.shape}")  # (rows, columns)
print(f"Columns: {df.columns.tolist()}")

print("\n3. DataFrame Data Types:")
print(df.dtypes)

print("\n4. Index:")
print(df.index)

print("\n5. Statistical Summary:")
print(df.describe())

print("\n6. First 3 rows:")
print(df.head(3))

print("\n7. Last 2 rows:")
print(df.tail(2))


# ============================================
# ALTERNATIVE METHOD: Using list of dictionaries
# ============================================
print("\n" + "=" * 50)
print("ALTERNATIVE: DataFrame from List of Dictionaries")
print("=" * 50)

data_list = [
    {'Name': 'Alice', 'Age': 25, 'City': 'New York'},
    {'Name': 'Bob', 'Age': 30, 'City': 'Paris'},
    {'Name': 'Charlie', 'Age': 28, 'City': 'Tokyo'},
    {'Name': 'Diana', 'Age': 35, 'City': 'London'},
]

df2 = pd.DataFrame(data_list)
print("\n")
print(df2)

print(f"\nType: {type(df2)}")
print(f"Shape: {df2.shape}")
```

### 4. Using .loc[] and .iloc[], show the first row, the second column, and a 2×2 block from the middle of your DataFrame.


```python
# dictionary list

data = {
    'Product': ['Pen', 'Book', 'Eraser', 'Pencil'],
    'Price': [10, 50, 5, 7],
    'Quantity': [100, 60, 200, 150],
    'Supplier' : ['Stationery Co', 'Book Mart', 'Office Supply', 'Stationery Co']
}

df = pd.DataFrame(data)
print(df)
```

```python
# show first row
print(df.iloc[0],'\n')
print(df.loc[0],'\n')
print('\n')

# show secound columns
print(df.iloc[:,1],'\n')
print(df.loc[:,'Price'],'\n')
```

```python
# 2×2 block from the middle of your DataFrame

print(df.iloc[1:3, 1:3])
```

### 5. Create a 4×3 DataFrame of student names and marks. Drop one column using .drop(), drop one row by index


```python
import pandas as pd

# Creating a 4x3 dataset
student_data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Math_Marks': [85, 92, 78, 95],
    'Science_Marks': [90, 88, 82, 89]
}

df = pd.DataFrame(student_data)
print(df)
```

```python
# Drop the 'Science_Marks' column
df.drop('Science_Marks', axis=1, inplace=True)

print(df)
```

```python
# Drop the row at index 2 (Charlie)
df.drop(2, axis=0, inplace=True)

print(df)
```

### 6. Create a DataFrame of employees with columns "Name", "Department", and "Salary". Sort it by "Salary" in ascending order and by "Name" in descending order

```python
import pandas as pd

# Setting up employee data
employee_data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Department': ['HR', 'Engineering', 'Marketing', 'Engineering', 'HR'],
    'Salary': [65000, 95000, 55000, 105000, 70000]
}

df = pd.DataFrame(employee_data)
print("Original DataFrame:")
print(df)
```

```python
print(df.sort_values(["Salary", "Name"], ascending=[0, 1]))
```

### 7. Create a DataFrame of 8 students with columns "Name" and "Score". Filter and print students with Score > 60 and those with Score < 40.

```python
import pandas as pd

# Creating a dataset with 8 students
student_scores = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva', 'Frank', 'Grace', 'Henry'],
    'Score': [50, 92, 35, 40, 70, 38, 91, 64]
}

# Load into a DataFrame
df = pd.DataFrame(student_scores)

print(df)
```

```python
# Filter Score > 60 and Score < 40

print(df.loc[(df['Score'] > 60) | (df['Score'] < 40)])

df
```

```python
# Fix: Use brackets () around each condition and & in between
print(df[(df['Score'] > 60) & (df['Score'] < 40)])
```
