## Filtering Based on string

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Eve", "Frank", "Grace", "Hannah"],
    "City": ["New York", "Los Angeles", "Newark", "Boston", "New Delhi", "Chicago", "New Orleans", "Hoston"],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Marketing", "Finance", "HR"],
    "Salary": [50000, 60000, 55000, 70000, 52000, 58000, 62000, 51000]
}

df = pd.DataFrame(data)
df
```

```python
df.loc[df['City'].str.contains("New")]
```

```python
df.loc[df['City'].str.contains("new", case=False)]
```

```python
# starts with los

df.loc[df["City"].str.contains(r"^Los", case=False)]
```

```python
# end with rk

df.loc[df['City'].str.contains(r"rk$", case=False)]
```

```python
# starts with A | B

df.loc[df["City"].str.contains(r"New|Los", case=False)]
```

## Handling Duplicates Values

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie", "David", "Alice", "Frank", "Grace", "Hannah", "Bob"],
    "City": ["New York", "Los Angeles", "Newark", "Boston","New York", "Chicago", "New Orleans", "Hoston", "Los Angeles"],
    "Department": ["HR", "IT", "Finance", "IT", "HR", "Marketing", "Finance", "HR", "IT"],
    "Salary": [50000, 60000, 55000, 70000, 52000, 58000, 62000, 51000, 60000]
}

df = pd.DataFrame(data)
df
```

```python
print(df.duplicated())
print(df.duplicated().sum())
```

```python
# deleting duplicates values

df.drop_duplicates()
```

```python
# permanent delete

df.drop_duplicates(inplace=True)
```

```python
df
```

```python
# name ar basis a duplicate deletion

df.drop_duplicates(subset=["Name"])
```

```python
df
```

```python
# name and city

df.drop_duplicates(subset=['Name','City'])
```

## Statistical Functions in Pandas

```python
import pandas as pd

df = pd.read_csv('student_data.csv')

df
```

```python
# sum
df['Data Structure Marks'].sum()
```

```python
# max and min
df.columns

print(df['Data Structure Marks'].min())
print(df['Data Structure Marks'].max())
```

```python
# mean -> average

df['Data Structure Marks'].mean()
```

```python
# median

df['Data Structure Marks'].median()
```

```python
# mode -> highest frequency number
df['Data Structure Marks'].mode()
```

```python
df
```

```python
# standard deviation -> std

# std = √(Σ(xi - mean)² / N)

# Where:
# - xi = each value
# - mean = average
# - N = total number of values
# - Σ = sum

df['Data Structure Marks'].std()

# STD RANGES: mean ± std coverage
# mean±1×std  → 68%  | mean±2×std  → 95%  | mean±3×std  → 99.7%
```

```python
# correlation matrix

df[['Data Structure Marks','Python Marks']].corr()
```

```python
# sum in different columns column wise

df[['Data Structure Marks','Python Marks']].sum()
```

```python
# sum in different columns row wise

df['Total Marks'] = df[['Data Structure Marks','Python Marks','Algorithm Marks']].sum(axis=1)

df
```

```python
# sum in different columns row wise

df['Total Marks'] = df['Data Structure Marks'] +df['Python Marks'] +df['Algorithm Marks']

df
```

```python
# sum in different columns row wise
df['Total Marks'] = df.iloc[:, 2:5].sum(axis=1)

df
```

```python
df.describe()
```

## apply function on df

```python
df
```

```python
# min max scaling

mn = df['Total Marks'].min()
mx = df['Total Marks'].max()

df['Scaled Marks'] = df['Total Marks'].apply(lambda x : (x-mn)/(mx-mn))
```

```python
df
```

```python
# custom built function

def grading_system(marks):

    if marks == 0:
        return 'F'
    elif marks >= 260:
        return 'A+'
    elif marks >= 250:
        return 'A'
    else:
        return 'A-'

df['Grade'] = df['Total Marks'].apply(grading_system)

df
```

```python
def marking_system(df):
    a = df['Data Structure Marks']*2
    b = df['Data Structure Marks']*3
    c = df['Data Structure Marks']*4
    return a+b+c
    
df['Exceptional Marks'] = df.apply(marking_system, axis=1)

df
```

## Pandas Group By

```python
import pandas as pd

df = pd.read_csv('student_completed_data.csv')

df
```

```python
import numpy as np

df.loc[df['FullName'] == 'Afsana Mimi', 'Total Marks'] = np.nan
df.loc[df['FullName'] == 'Sadia Chowdhury', 'Total Marks'] = np.nan
df.loc[df['FullName'] == 'Alif Rahman', 'Total Marks'] = np.nan

df
```

```python
instructior_group = df.groupby('Instructor')

instructior_group
```

```python
for instructior, df_ins in instructior_group: 
    print(instructior)
    display(df_ins)
    print('\n')
```

```python
df.min()
```

```python
instructior_group.min() 
```

```python
instructior_group.describe()
```

```python
df.info()
```

```python
min_student = df[df['Total Marks'] == df['Total Marks'].min()]
print(min_student)
```

```python
min_student = df.loc[df['Total Marks'].idxmin()]
print(min_student)
```

```python
instructior_group['Total Marks'].mean()
```

```python
mean_marks = instructior_group['Total Marks'].mean()

mean_marks
```

```python
df['Total Marks']=df['Total Marks'].fillna(df['Instructor'].map(mean_marks))

df
```
