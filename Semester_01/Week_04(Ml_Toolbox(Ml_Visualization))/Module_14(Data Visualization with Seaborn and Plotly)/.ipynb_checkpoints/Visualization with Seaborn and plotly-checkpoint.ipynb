```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
student = pd.read_csv('sns_data.csv') 
student.head()
```

## Lineplot using Seaborn 

```python
sns.lineplot(student, x='week',y='attendance_rate')
```

```python
sns.lineplot(data=student,y='attendance_rate', x='week')
```

```python
# Axes label
sns.lineplot(data=student, x='week',y='attendance_rate', errorbar=None, hue='gender')
```

```python
# Figure label (relplot = relational plot)
sns.relplot(kind='line',
            data=student,
            x='week',
            y='attendance_rate',
            errorbar=None,
            hue='gender'
           )
```

```python
sns.relplot(kind='line',
            data=student,
            x='week',
            y='test_score',
            errorbar=None,
            hue='class_level'
           )
```

```python
sns.relplot(kind='line',
            data=student,
            x='week',
            y='attendance_rate',
            errorbar=None,
            hue='class_level'
           )
```

```python
fig, ax1 = plt.subplots(figsize=(8, 5))

# First y-axis (Attendance Rate)
sns.lineplot(data=student, x='week', y='attendance_rate', hue='gender', ax=ax1, errorbar=None)
ax1.set_ylabel('Attendance Rate')

# Second y-axis (Test Score)
ax2 = ax1.twinx()
sns.lineplot(data=student, x='week', y='test_score', hue='gender', ax=ax2, linestyle='--', errorbar=None, legend=False)
ax2.set_ylabel('Test Score')

plt.title('Attendance Rate and Test Score by Week')
plt.show()
```

## Scatter plot using Seaborn

```python
#positive co-relation
plt.figure(figsize=(7, 6.5))
sns.scatterplot(data=student, 
                x='study_hours', 
                y='test_score', 
                hue='gender', 
                style='class_level',
                size='Tshirt_size'
               )
plt.legend(loc='upper left')
```

```python
#positive co-relation
sns.relplot(kind='scatter',
                data=student, 
                x='study_hours', 
                y='test_score', 
                hue='gender', 
                style='class_level',
                size='Tshirt_size'
               )
```

```python
tips_data = sns.load_dataset('tips')

tips_data.head()
```

```python
sns.relplot(kind='scatter',
           data=tips_data,
            x='total_bill', 
            y='tip',
            hue='sex', 
            style='time', 
            size='size'
           )
```

## Facet Plot 

```python
student = pd.read_csv('student_dataset_complete.csv')

student.head()
```

```python
sns.scatterplot(data= student, x='study_hours', y='test_score', hue='gender')
```

```python
sns.relplot(kind='scatter',
            data= student,
            x='study_hours',
            y='test_score',
            col='gender',
            row='hostel'
           )
```

```python
sns.relplot(kind='scatter',
            data= student,
            x='study_hours',
            y='test_score',
            col='week',
            col_wrap=2
           )
```

## histogram using Seaborn

```python
# axes level

sns.histplot(data=student, x='attendance_rate', hue='gender', bins=15, element='step')
```

```python
# figure level displot(distribution plot)

sns.displot(kind='hist', data=student, x='attendance_rate', col='gender')
```

## KDE plot using Seaborn

```python
# KDE(karnel density etimation) plot using seaborn
sns.kdeplot(data=student, x='attendance_rate')
```

```python
sns.kdeplot(data=tips_data, x='tip', hue='sex', fill=True)
```

```python
sns.displot(kind='kde',data=tips_data, x='tip', col='sex', fill=True)
```

## Countplot

```python
sns.countplot(data=student, x='gender')
```

```python
sns.countplot(data=student, x='subject', hue='gender')
```

## Bar plot

```python
sns.barplot(data=student, x='gender', y='Marks', errorbar=None) # mean
```

```python
sns.barplot(data=student, x='gender', y='Marks', errorbar=None, estimator=np.median) # median
```

```python
sns.barplot(data=student, x='gender', y='Marks', errorbar=None, estimator=np.max) # max
```

## Regplot and lmplot (OPTIONAL) 

```python
students = pd.read_csv('student_dataset_complete.csv')
```

```python
# Regplot(Regression plot)
# axes level
sns.regplot(data = students, x='study_hours', y='test_score')
```

```python
# fegure level
sns.lmplot(data = students, x='study_hours', y='test_score', hue='gender')
```
