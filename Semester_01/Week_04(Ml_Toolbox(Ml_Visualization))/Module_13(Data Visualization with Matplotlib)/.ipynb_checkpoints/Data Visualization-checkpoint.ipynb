```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
```

## 2D line plot

```python
hours = [2, 3, 4, 1, 5, 7, 3]
days = [1, 2, 3, 4, 5, 6, 7]

# plt.plot(x,y)
plt.plot(days, hours)
```

```python
plt.xlabel('Days')
plt.ylabel('Hours')
plt.title("Trend a student study time over a week")

plt.plot(days, hours)
```

```python
hours_1 = [2, 3, 4, 1, 5, 7, 3]
hours_2 = [1, 4, 3, 5, 7, 2, 1]
days = [1, 2, 3, 4, 5, 6, 7]

plt.plot(days, hours_1)
plt.plot(days, hours_2)
```

## 2D line plot from pandas data

```python
df = pd.read_csv('enrollment_data.csv')
df
```

```python
plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student ENrollment over 8 years')

plt.plot(df['Year'], df['Programming'])
```

```python
plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student ENrollment over 8 years')

plt.plot(df['Year'], df['Programming'])
plt.plot(df['Year'], df['Digital Marketing'])
```

```python
# adding labels using plt.legend()

plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student ENrollment over 8 years')

plt.plot(df['Year'], df['Programming'], label='Programming-1')
plt.plot(df['Year'], df['Digital Marketing'], label='Digital Marketing')

plt.legend()
```

```python
# adding labels using plt.legend(loc='lower right')

plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student ENrollment over 8 years')

plt.plot(df['Year'], df['Programming'], label='Programming')
plt.plot(df['Year'], df['Digital Marketing'], label='Digital Marketing')

plt.legend(loc = 'lower right')
```

```python
# adding labels using plt.legend('symbol_1symbol_2',loc='lower right')

plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student ENrollment over 8 years')

plt.plot(df['Year'], df['Programming'], label='Programming')
plt.plot(df['Year'], df['Digital Marketing'], label='Digital Marketing')

plt.legend('PD',loc = 'lower right')
```

```python
# adding some visualization

plt.xlabel('Years')
plt.ylabel('Count of Enrollment')
plt.title('Student ENrollment over 8 years')

plt.plot(df['Year'], df['Programming'], label='Programming', color ='red', linestyle ='dashed', marker ='o')
plt.plot(df['Year'], df['Digital Marketing'], label='Digital Marketing', color = '#4deeea', linewidth=5, linestyle='dashdot', marker ='o', markersize='10')

plt.legend()
plt.grid()
plt.show()
```

## Scatter Plot

```python
students = pd.read_csv('student_IQdata.csv')

students
```

```python
# Positive co-relation
# Relation betweeen Study Hours and IQ Score

plt.xlabel('Study hours')
plt.ylabel('IQ Score')
plt.title('Relation betweeen Study Hours and IQ Score')

plt.scatter(students['Study_Hour'], students['IQ_Score'], marker='*')
# Here 2 hour study = 145 Iq and 18 hour study = 75 Iq this twois Outlier
```

```python
# No Co-relation
# Relation betweeen Shoe Size and IQ Score

plt.xlabel('Shoe Size')
plt.ylabel('IQ Score')
plt.title('Relation betweeen Shoe Size and IQ Score')

plt.scatter(students['Shoe_Size'], students['IQ_Score'])
```

```python
# Negative co-relation
# Relation betweeen Chilling_Hours and IQ Score

plt.xlabel('Chilling Hours')
plt.ylabel('IQ Score')
plt.title('Relation betweeen Chilling Hours and IQ Score')

plt.scatter(students['Chilling_Hours'], students['IQ_Score'], color = 'red')
# Here 35 hour chilling = 145 Iq this twois Outlier
```

## Histogram

```python
#Based on IQ
plt.xlabel('IQ Score')
plt.ylabel('Frequency of Students')
plt.hist(students['IQ_Score'], color='white', edgecolor='#4deeea')
```

```python
#Based on Study Hours

plt.xlabel('Study_Hour')
plt.ylabel('Frequency of Students')
plt.hist(students['Study_Hour'], color='#4deeea', edgecolor='black')
```

```python
#Based on Chilling Hours

plt.xlabel('Chilling Hours')
plt.ylabel('Frequency of Students')
plt.hist(students['Chilling_Hours'], color='#4deeea', edgecolor='black', bins=[5.0, 6.9, 10.0, 14.0, 17.7, 20.9,  22.8, 24.3, 27.2, 30, 35.0])
```

## Bar Chart

```python
data = {
    'Student' : ['A', 'B', 'C', 'D', 'E', 'F', 'G'],
    'Gender' : ['Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'StudyHours' : [3, 5, 3, 7, 4, 8, 1], 
    'Attendence' : [92, 88, 70, 95, 80, 98, 55],
    'Grade' : ['Pass', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
}

df = pd.DataFrame(data)

df
```

```python
# gender wise

gender_group = df.groupby('Gender').size()
type(gender_group)
```

```python
gender_group.index
```

```python
gender_group.values
```

```python
plt.xlabel('Gender')
plt.ylabel('Count')
plt.title('Bar Chart of Gender among students')

plt.bar(gender_group.index, gender_group.values, color = ['#4deeea', 'yellow'], edgecolor = 'black')
```
